```python
def create_sine_signal(M, A=1.0, phi= 0.0, fs=48e3, dur=1.0):
    '''
    M  : period (samples)
    A  : amplitude
    phi: phase
    dur: duration (seconds)
    fs : sampling frequency (Hz)
    '''

    M = int(M)
    w0 = 2 * np.pi / M
    phi = 2 * np.pi / 3
    f = fs / M # frequencia do sinal (Hz)
    N = int(fs * dur)
    n = np.arange(N)

    x = A * np.sin(w0 * n + phi)

    return x, n
```
